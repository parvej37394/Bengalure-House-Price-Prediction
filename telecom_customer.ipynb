```python
import pandas as pd
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
data = pd.read_csv("Telco-Customer-Churnn.csv")
```

# 1. Does Churn Rate Differ Between Male and Female Customers?

```python
churn_gender = pd.crosstab(data['gender'],data['Churn'],normalize='index') * 100

print(churn_gender)

churn_gender.plot(
    kind='bar',
    figsize=(8, 5)
)

plt.title('Churn Rate by Gender')
plt.xlabel('Gender')
plt.ylabel('Churn Percentage')
plt.xticks(rotation=0)
plt.legend(title='Churn')
plt.show()
```

# 2. Which Contract Type Has the Highest Customer Churn Rate?

```python
contact_churn = pd.crosstab(data['Contract'],data['Churn'], normalize = 'index')*100
print(contact_churn)

contact_churn.plot(kind ='bar',figsize =(8,5))
plt.title("Churn Rate by contract type")
plt.xlabel("Contract Type")
plt.ylabel("Churn Percentage")
plt.xticks(rotation = 10,color='red')
plt.legend(title ='Churn')
plt.show()
```

# 3. Does a Higher Monthly Charge Lead to a Higher Customer Churn Rate?

```python
plt.figure(figsize=(8, 5))

sns.boxplot(
    x='Churn',
    y='MonthlyCharges',
    data=data
)

plt.title('Monthly Charges vs Customer Churn')
plt.xlabel('Churn')
plt.ylabel('Monthly Charges')

plt.show()
```

```python
print(
    data.groupby('Churn')['MonthlyCharges'].mean()
)
```

# 4. How Does Customer Tenure Affect the Churn Rate?'

```python
data['tenure_group'] = pd.cut(
    data['tenure'],
    bins=[0, 12, 24, 48, 72],
    labels=[
        '0-12 Months',
        '13-24 Months',
        '25-48 Months',
        '49-72 Months'
    ],
    include_lowest=True
)

tenure_churn = pd.crosstab(
    data['tenure_group'],
    data['Churn'],
    normalize='index'
) * 100

print(tenure_churn)

tenure_churn.plot(
    kind='bar',
    figsize=(9, 5)
)

plt.title('Churn Rate by Customer Tenure')
plt.xlabel('Customer Tenure')
plt.ylabel('Churn Percentage')
plt.xticks(rotation=0)
plt.legend(title='Churn')
plt.show()
```
